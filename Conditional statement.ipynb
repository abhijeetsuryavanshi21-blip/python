# Conditional Statement------>

Conditional statement is used to control the flow of the program.

# Python offers three types of conditional statements

1. if statement ---> When we have only one condition then use if.
2. if-else statement ---> When we have two conition then use if-else.
3. if-elif-else statement ---> When we have more than two conditions then we use if-elif-else.

All the conditional statement will work when all the conditions are true

syntax ----> if......Condition

# Q.13 :- WAP to print week name if user enter week number.

```python
day= int(input('Enter a number of day----> '))
if day>0 and day<=7:
  if day==1:
    print('Its Monday')
  elif day==2:
    print('Its Tuesday')
  elif day==3:
    print('Its Wednesday')
  elif day==4:
    print('Its Thusday')
  elif day ==5:
    print('Its Friday')
  elif day==6:
    print('Its Saturday')
  else:
    print("Its Sunday")
else:
  print('Invalid Input')
```

## Q.14 :-  Write a program to find area of rectangle or square take dimension from user?

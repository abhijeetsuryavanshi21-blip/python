Label Encoder

---> Label Encoding converts categorical values into integer labels, mainly for representing categories numerically. the assigned numbers do not necessarily represent any meaningful order.

```python
import pandas as pd 
import seaborn as sns
```

```python
df= sns.load_dataset('tips')
```

```python
from sklearn.preprocessing import LabelEncoder
encoder_time= LabelEncoder()
encoded_data= encoder_time.fit_transform(df['time'])
```

```python
encoded_data
```

```python
df
```

```python
print("classes", encoder_time.classes_)
```

```python
for category, number in zip (encoder_time.classes_, range(len(encoder_time.classes_))):
    print(category, "--------", number)
```

```python
encoder_day= LabelEncoder()

encoded_day_data= encoder_day.fit_transform(df['day'])
```

```python
print("classes", encoder_day.classes_)
```

```python
for category, number in zip (encoder_day.classes_, range(len(encoder_day.classes_))):
    print(category, "--------", number)
```

```python
encoder_day.inverse_transform([0,1,2,3])
```

One-Hot Encoder

---> One-Hot Encoding converts a categorical feature into multiple binary (0/1) columns, with one column for each category.

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
ohe=OneHotEncoder(sparse_output=False)

ohe.fit_transform(df[['sex','smoker',]])
```

```python
ohe=OneHotEncoder(sparse_output=False, drop='first')

ohe_encoder=ohe.fit_transform(df[['sex','smoker',]])
ohe_encoder
```

```python
df1=pd.DataFrame(ohe_encoder,columns=ohe.get_feature_names_out())
df1
```

```python
pd.concat([df,df1],axis=1)
```

Function Transformer

---->FunctionTransformer is a scikit-learn transformer that allows you to apply a custom Python function to input features while keeping that transformation compatible with the scikit-learn Transformer API.

```python
from sklearn.preprocessing import FunctionTransformer
import numpy as np
```

```python
log=FunctionTransformer(np.log10)
log.fit_transform(df['tip'])
```

```python
log_2=FunctionTransformer(np.log2)
log_2.fit_transform(df['tip'])
```

Power Transformer

```python
import numpy as np
from sklearn.preprocessing import PowerTransformer

X = np.array([
    [-1],
    [0],
    [3],[5],[10],[50]
])
pt= PowerTransformer(method="yeo-johnson")

X_transformed = pt.fit_transform(X)

print('original')
print(X.ravel())

print('\nTransformed:')
print(np.round(X_transformed.ravel(),3))
print('\nLearned lambda:')
print(pt.lambdas_)
```

```python
Y = np.array([
    [1],
    [1],
    [3],[5],[10],[50]
])
pt= PowerTransformer(method="box-cox")

Y_transformed = pt.fit_transform(Y)

print('original')
print(Y.ravel())

print('\nTransformed:')
print(np.round(X_transformed.ravel(),3))
print('\nLearned lambda:')
print(pt.lambdas_)
```

Missing-Value Strategies

---> Missing values are values that are not present in a dataset. They are usually represented as NaN, None, or blank values.

```python
import numpy as np
import pandas as pd

# Creating a sample dataset
data = {
    "Name": [
        "Aarav",
        "Priya",
        "Rohan",
        "Sneha",
        "Vikram",
        "Ananya",
        "Karan",
        "Pooja",
        "Rahul",
        "Neha",
    ],
    "Age": [25, np.nan, 30, 22, np.nan, 28, 35, np.nan, 40, 29],  # Numerical
    "Salary": [
        50000,
        64000,
        np.nan,
        45000,
        80000,
        np.nan,
        95000,
        52000,
        np.nan,
        61000,
    ],  # Numerical
    "Experience_Years": [
        2,
        np.nan,
        5,
        1,
        8,
        3,
        np.nan,
        2,
        12,
        4,
    ],  # Numerical
    "Department": [
        "IT",
        "HR",
        "IT",
        np.nan,
        "Finance",
        "Marketing",
        "Finance",
        np.nan,
        "IT",
        "HR",
    ],  # Categorical
    "City": [
        "Mumbai",
        "Pune",
        np.nan,
        "Delhi",
        "Mumbai",
        np.nan,
        "Bangalore",
        "Pune",
        "Delhi",
        "Mumbai",
    ],  # Categorical
    "Purchased": [
        "No",
        "Yes",
        "Yes",
        "No",
        np.nan,
        "Yes",
        "No",
        "Yes",
        "No",
        np.nan,
    ],  # Categorical
}

df = pd.DataFrame(data)

# Display data
display(df)

# Check missing values count per column
print("\nMissing values per column:")
print(df.isnull().sum())

# Check data types and non-null counts
print("\nData info:")
df.info()
```

```python
from sklearn.impute import SimpleImputer
```

1. Mean Imputation

```python
imputer= SimpleImputer(strategy='mean')
imputer.fit_transform(df[['Salary']])
```

```python
imputer= SimpleImputer(strategy='mean')
df['new_salary']= imputer.fit_transform(df[['Salary']]).flatten()
```

2. Median Imputation

```python
imputer= SimpleImputer(strategy='median')
df['new_Experience_Years']= imputer.fit_transform(df[['Experience_Years']]).flatten()
```

3. Mode Imputation

```python
imputer= SimpleImputer(strategy='most_frequent')
df['City']= imputer.fit_transform(df[['City']]).flatten()
```

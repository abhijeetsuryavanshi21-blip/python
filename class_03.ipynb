```python
array = np.linspace(12,22,num=15,dtype=int)
print(array)
```

```python
def variance (*args):
    total=0
    for i in args:
        total+=(i-np.mean(args))**2
    calculated_variance = total/len(args)
    return calculated_variance, calculated_variance**(1/2)
```

```python
variance(*array)
```

```python
def variance (*args):
    total=0
    for i in args:
        total+=(i-np.mean(args))**2
    calculated_variance = total/(len(args)-1)
    return calculated_variance, calculated_variance**(1/2)
```

```python
import matplotlib.pyplot as plt
```

```python
hours = [1,2,3,4,5,6]
marks = [30,40,45,55,65,75]
```

```python
plt.Figure(figsize=(3,3))
plt.scatter(hours,marks)
plt.title("Hours vs Marks")
plt.xlabel("hours")
plt.ylabel("Marks")
plt.show()
```

```python
Products = ["laptop","mobile","tablet","headphone"]
Sales = [40,80,30,60]
plt.Figure(figsize=(3,3))
plt.bar(Products , Sales)
plt.title("Products vs Sales")
plt.xlabel("<------Products------->")
plt.ylabel("<------Sales------->")
plt.tight_layout()
plt.show()
```

# **4. Logical Operators**

AND Operator----> When all conditions are True the output is True otherwise False.

```python
x = 50
y = 80
result = (x>0) and (y>0)
print(result)
```

```python
x = 50
y = 80
result = (x>y) and (y>0)
print(result)
```

OR Operator----> When atleast one condition is True than output is True otherwise False.

```python
x = 50
y = 80
result = (x>y) or (y>0)
print(result)
```

```python
x = 50
y = 80
result = (x>y) and (y>0)
print(result)
```

# **6. Membership (in) Operator**
Membership operator are used to check whether a value exists in a sequence such as a list, string, tuple, or set.

```python
name= "Abhijeet"
print("jeet" in name)
```

```python
'a' in 'abc'
```

```python
fruits = ("Apple","Orange","Banana")
print("Apple"in fruits)
```

# Q.6:- Write a program that take a number from user then add all natural number which are divisible by 3 or 5 both till input number.


```python
total=0
count=0
while True:
  num = int(input('enter a num '))
  if num % 3 == 0 or num % 5 == 0:
    break
  total+=num
  count+=1
  print(num)
print(f'Total: {total}, Count: {count}')
```

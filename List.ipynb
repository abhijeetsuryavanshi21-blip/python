# Q.1 :- WAP to store five element in the list using while loop.

```python
empty_lst = []
i=0
while i<5:
  num= int(input('Enter a number'))
  empty_lst.append(num)
  i+=1
  print(empty_lst)
```

# Q.2 :- WAP to find average of the number of list. List should be contain 7 number. Number should be given by user.

```python
empty_lst = []
i=0
while i<7:
  num= int(input('Enter a number'))
  empty_lst.append(num)
  i+=1
print(empty_lst)

total=0
for i in empty_lst:
  total+=i
print(total/len(empty_lst))
```

# Q.3 :- [5,7,8,'Hello','Ram',18,21,Sita] Create two list first for numbers and another for string.

```python
list3=[5,7,8,'Hello','Ram',18,21,"Sita"]
lst_num=[]
lst_str=[]
for i in list3:
  if type(i)==int:
    lst_num.append(i)
  else:
    lst_str.append(i)

print("List1--->",lst_num)
print("List2--->",lst_str)
```

# Q.6 :- WAP to find prime number for any one number.


```python
n=int(input("Enter a number: "))
for i in range(2,n):
  if n%i==0:
    print("Not prime number")
    break
  else:
    print("Prime number")
```

# Q.7 :- WAP to find perfect number for any one number.


```python
n = int(input('Enter a number: '))
total=0
for i in range(1,n):
  if n%i==0:
    total+=i
if total==n:
  print("Perfect Number")
else:
  print("Not Perfect Number")
```

# Q.9 :- WAP to find factorial number from 1 to 100.

```python
for i in range(1,101):
  fact=1
  for i in range(1,i+1):
    fact*=i
  print("Factorial of",i,"=",fact)
```

# Q.10 :- WAP to find perfect number from 1 to 100,

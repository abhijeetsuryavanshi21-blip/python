# Dictionary
Dictionary is a data container which store the value in key value pair.

{key:value}

We use "{}" carlly bracket to store the data in dictionary.

It's Properties---->

1. Unordered ----> No index position
2. Mutable
3. Accept only unique keys.



```python
dict1={'Abhijeet',20,9.55}
```

```python
len(dict1)# Return number of keys
```

```python
dict2={'name':'Abhijeet',
       'college':'JNCT',
       'CGPA':9.55}
```

```python
dict2.keys() # Store all the keys.
```

```python
dict2.values() # Store all the values of keys.
```

```python
dict2.items() # Items store both keys and his values.
```

```python
dict2.pop('college') # Pop function are return or remove the value.
```

```python
dict2.update({"place":["Microsoft",'Cisco']}) # update dict by dict.
```

```python
dict2
```

```python
dict2.get('name') # it returns relative value of keys.
```

```python
dict2['marks']=[9.5, 8.9, 8.5, 9.2]
```

```python
dict2
```

```python
dict2['sub']=['ML', 'C0A', 'SE', 'IDS']
```

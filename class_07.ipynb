Simple Linear Regression
---> to find relationship b\t input and output variables
----> simple linear regression means one independent and one denpendent variable

```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt 
from sklearn.linear_model import LinearRegression
```

```python
x = np.array([1,2,3,4,5,6,7,8])
y = np.array([12,22,32,42,52,62,72,80])
```

```python
x.ndim
```

```python
x = np.array([1,2,3,4,5,6,7,8]).reshape(-1,1)
y = np.array([12,22,32,42,52,62,72,80])
```

```python
x.ndim
```

```python
model= LinearRegression()
model.fit(x,y)
```

```python
m=model.coef_
c= model.intercept_
```

```python
y_pred=model.predict(x)
```

Q1. Employee Salary Analysis

A company records the monthly salaries of its employees, in thousands of rupees.

salaries = [18, 20, 22, 22, 24, 25, 25, 26, 28, 30, 32, 35, 38, 42, 150]

(a) Write Python functions to calculate the mean, median, and all modes without using statistics.mean(), statistics.median(), or statistics.mode(). Display the results for the original dataset.

(b) Remove the salary of 150 and recalculate all three measures. Calculate the percentage change in the mean and median, then justify which measure better represents a typical employee's salary.

```python
salaries = [18, 20, 22, 22, 24, 25, 25, 26, 28, 30, 32, 35, 38, 42, 150]

def calculate_mean(data): 
    return sum(data)/len(data)
def calculate_median(data):
    data= sorted(data)
    n=len(data)
    if n%2==1:
        return data[n//2]
    else:
        return (data[n//2-1]+ data[n//2])/2
def calculate_modes(data):
    frequency={}
    for value in data:
        frequency[value]=frequency.get(value,0)+1
    max_frequency = max(frequency.values())
    modes=[]
    for value, count in frequency.items():
        if count==max_frequency and count >1:
            modes.append(value)
    return modes
print("Mean---",calculate_mean(salaries))
print('Median---',calculate_median(salaries))
print("Mode---",calculate_modes(salaries))
```

```python
new_salaries = [x for x in salaries if x != 150]
old_mean=calculate_mean(salaries)
new_mean = calculate_mean(new_salaries)

old_median= calculate_median(salaries)
new_median= calculate_median(new_salaries)

print("New Mean---",new_mean)
print("new median---",new_median)
print("new mode---",calculate_modes(new_salaries))

mean_change=((new_mean - old_mean)/old_mean)*100
median_change=((new_median - old_median)/old_median)*100

print("mean percentage change--",round(mean_change,2),"%")
print("median percentage change--",round(median_change,2),"%")
```

Q3. Build a Central Tendency Calculator From Scratch

Use the following datasets:

data_1 = [12, 15, 15, 18, 20, 22, 25]
data_2 = [5, 5, 10, 10, 15, 20]
data_3 = [7, 14, 21, 28, 35]
data_4 = [100, 100, 100, 100]
data_5 = [1, 2, 3, 4, 5, 100]

(a) Build reusable Python functions to calculate the mean, median, and mode without using built-in statistical functions. Your program must handle unsorted data, odd and even dataset lengths, multiple modes, and datasets with no repeated values.

(b) Run your calculator on all five datasets and generate a formatted report. Identify which dataset is most affected by an extreme value and demonstrate numerically why its mean and median behave differently.

```python
data_1 = [12, 15, 15, 18, 20, 22, 25]
data_2 = [5, 5, 10, 10, 15, 20]
data_3 = [7, 14, 21, 28, 35]
data_4 = [100, 100, 100, 100]
data_5 = [1, 2, 3, 4, 5, 100]
```

```python
def calculate_mean(data):
    total = 0
    for value in data:
        total += value
    return total / len(data)

def calculate_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)
    if n % 2 == 1:
        return sorted_data[n // 2]
    else:
        middle1 = sorted_data[n // 2 - 1]
        middle2 = sorted_data[n // 2]
        return (middle1 + middle2) / 2

def calculate_mode(data):
    frequency = {}
    for value in data:
        frequency[value] = frequency.get(value, 0) + 1
    max_frequency = max(frequency.values())
    if max_frequency == 1:
        return []
    modes = []
    for value, count in frequency.items():
        if count == max_frequency:
            modes.append(value)
    return modes
print("functions are defined successfully.")
```

```python
datasets = {"Data 1": data_1,
    "Data 2": data_2,
    "Data 3": data_3,
    "Data 4": data_4,
    "Data 5": data_5}
print("=" * 55)
print("CENTRAL TENDENCY ANALYSIS REPORT")
print("=" * 55)

for name, data in datasets.items():
    print("\n", name)
    print("Dataset:", data)
    print("Mean:", round(calculate_mean(data), 2))
    print("Median:", calculate_median(data))
    print("Mode:", calculate_mode(data))

print("\n" + "=" * 55)
print("EXTREME VALUE ANALYSIS: DATA 5")
print("=" * 55)

mean_before = calculate_mean(data_5)
median_before = calculate_median(data_5)
data_5_without_outlier = [x for x in data_5 if x != 100]
mean_after = calculate_mean(data_5_without_outlier)
median_after = calculate_median(data_5_without_outlier)

print("Original Data:", data_5)
print("Mean before removal:", round(mean_before, 2))
print("Median before removal:", median_before)
print("Data after removing 100:", data_5_without_outlier)
print("Mean after removal:", round(mean_after, 2))
print("Median after removal:", median_after)
print("Mean Change:", round(mean_before - mean_after, 2))
print("Median Change:", round(median_before - median_after, 2))
```

Q4. Statistical Transformations and Reasoning

An online store records the number of products purchased per order.

orders = [1, 1, 1, 2, 2, 2, 2, 3, 3, 4, 4, 5, 5, 6, 8, 10, 15, 20, 25, 80]

(a) Calculate the mean, median, and all modes. Remove the largest observation and recalculate the measures. Use the results to explain how an extreme order affects each measure.

(b) Create two transformed datasets: add 20 to every observation in the first transformation and multiply every observation by 3 in the second. Calculate the mean, median, and mode for each dataset, then use Python to verify how these transformations affect the original measures.

```python
orders = [1, 1, 1, 2, 2, 2, 2, 3, 3, 4,4, 5, 5, 6, 8, 10, 15, 20, 25, 80]
```

```python
def calculate_mean(data):
    return sum(data) / len(data)

def calculate_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)
    if n % 2 == 1:
        return sorted_data[n // 2]
    else:
        return (sorted_data[n // 2 - 1] + sorted_data[n // 2]) / 2

def calculate_modes(data):
    frequency = {}
    for value in data:
        frequency[value] = frequency.get(value, 0) + 1
    max_frequency = max(frequency.values())
    if max_frequency == 1:
        return []
    return [value for value, count in frequency.items()
            if count == max_frequency]

print("ORIGINAL DATA")
print("Mean:", calculate_mean(orders))
print("Median:", calculate_median(orders))
print("Modes:", calculate_modes(orders))
new_orders = orders.copy()
new_orders.remove(max(new_orders))
print("AFTER REMOVING LARGEST OBSERVATION")
print("New Data:", new_orders)
print("Mean:", calculate_mean(new_orders))
print("Median:", calculate_median(new_orders))
print("Modes:", calculate_modes(new_orders))
```

```python
original_mean = calculate_mean(orders)
original_median = calculate_median(orders)
original_mode = calculate_modes(orders)
add_20 = [x + 20 for x in orders]
multiply_3 = [x * 3 for x in orders]

print("ORIGINAL DATA")
print("Mean:", original_mean)
print("Median:", original_median)
print("Mode:", original_mode)

print("\nAFTER ADDING 20")
print("Mean:", calculate_mean(add_20))
print("Median:", calculate_median(add_20))
print("Mode:", calculate_modes(add_20))

print("\nAFTER MULTIPLYING BY 3")
print("Mean:", calculate_mean(multiply_3))
print("Median:", calculate_median(multiply_3))
print("Mode:", calculate_modes(multiply_3))
```

Q5. Missing Values and Outlier Investigation

A data analyst receives the following daily sales figures, in thousands of rupees.

sales = [12, 15, 14, 16, None, 15, 18, 14, 17, 16, None, 19, 15, 120, 14]

(a) Convert the data into a Pandas Series, identify missing observations, and calculate the percentage of missing values. Excluding missing values, calculate the mean, median, and all modes. Then use the IQR method to identify potential outliers.

(b) Recalculate the mean, median, and mode after excluding the suspected outlier of 120. Compare the results, calculate the percentage changes in the mean and median, and write a Python-supported recommendation on whether the outlier should be removed, retained, or investigated further.

```python
import pandas as pd
```

```python
sales = [12, 15, 14, 16, None, 15, 18, 14, 17, 16, None, 19, 15, 120, 14]
```

```python
sales_series = pd.Series(sales)
missing_values = sales_series.isnull()
print("Missing Observations:")
print(sales_series[missing_values])
missing_percentage = (sales_series.isnull().sum() /len(sales_series)) * 100
print("Number of Missing Values:", sales_series.isnull().sum())
print("Percentage of Missing Values:",round(missing_percentage, 2), "%")
clean_sales = sales_series.dropna()

def calculate_modes(data):
    frequency = {}
    for value in data:
        frequency[value] = frequency.get(value, 0) + 1
    max_frequency = max(frequency.values())
    return [value for value, count in frequency.items()
            if count == max_frequency]

print("STATISTICAL MEASURES")
print("Mean:", clean_sales.mean())
print("Median:", clean_sales.median())
print("Modes:", calculate_modes(clean_sales.tolist()))

Q1 = clean_sales.quantile(0.25)
Q3 = clean_sales.quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("IQR ANALYSIS")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

outliers = clean_sales[(clean_sales < lower_bound) | (clean_sales > upper_bound)]

print("Potential Outliers:")
print(outliers)
```

```python
original_mean = clean_sales.mean()
original_median = clean_sales.median()
original_modes = calculate_modes(clean_sales.tolist())
sales_without_outlier = clean_sales[clean_sales != 120]
new_mean = sales_without_outlier.mean()
new_median = sales_without_outlier.median()
new_modes = calculate_modes(sales_without_outlier.tolist())

print("BEFORE REMOVING OUTLIER")
print("Mean:", original_mean)
print("Median:", original_median)
print("Modes:", original_modes)

print("AFTER REMOVING OUTLIER")
print("Mean:", new_mean)
print("Median:", new_median)
print("Modes:", new_modes)

mean_change = ((new_mean - original_mean) /original_mean) * 100

median_change = ((new_median - original_median) / original_median) * 100

print("PERCENTAGE CHANGE")
print("Mean Change:", round(mean_change, 2), "%")
print("Median Change:", round(median_change, 2), "%")
```

Q1. Employee Performance: Population vs. Sample Variance
A company records the performance scores of eight employees.

scores = [45, 50, 52, 55, 58, 60, 65, 75]
(a) Write Python code to calculate the mean, population variance, sample variance, population standard deviation, and sample standard deviation. Implement the variance and standard deviation formulas manually without using np.var() or np.std() for the calculations.

(b) Explain why the population and sample variance differ. Quantify the difference using the calculated results and recommend which variance formula should be used if these eight employees represent a sample of a much larger workforce.

```python
scores = [45, 50, 52, 55, 58, 60, 65, 75]
```

```python
n = len(scores)
mean = sum(scores) / n
squared_deviations = []
for score in scores:
    deviation = score - mean
    squared_deviations.append(deviation ** 2)
population_variance = sum(squared_deviations) / n
sample_variance = sum(squared_deviations) / (n - 1)
population_std = population_variance ** 0.5
sample_std = sample_variance ** 0.5

print("Mean:", mean)
print("Population Variance:", round(population_variance, 2))
print("Sample Variance:", round(sample_variance, 2))
print("Population Standard Deviation:", round(population_std, 2))
print("Sample Standard Deviation:", round(sample_std, 2))
```

Ans--(b) 

Population variance divides the sum of squared deviations by the total number of observations,n. Sample variance divides by n-1, which is called Bessel's correction.
If these eight employees represent a sample of a much larger workforce, sample variance should generally be used to estimate the workforce's variance.
This is because dividing by n-1 corrects for the tendency of variance calculated around a sample mean to underestimate the true population variance.
Population variance is 77.25, while sample variance is 88.29. The difference occurs because population variance divides the sum of squared deviations by n, whereas sample variance divides it by n-1. This correction helps reduce the underestimation of population variance when working with a sample. Therefore, if the eight employees represent a sample of a larger workforce, sample variance and sample standard deviation are more appropriate for estimating the population's variability.

```python
import seaborn as sns
import pandas as pd
```

```python
tips = sns.load_dataset("tips")
```

```python
tips
```

```python
for col in ["total_bill", "tip"]:
    data = tips[col]

    mean = sum(data) / len(data)

    variance = sum((x - mean) ** 2 for x in data) / len(data)

    std_dev = variance ** 0.5

    cv = (std_dev / mean) * 100

    print("\nColumn:", col)
    print("Mean:", round(mean, 2))
    print("Variance:", round(variance, 2))
    print("Standard Deviation:", round(std_dev, 2))
    print("Coefficient of Variation:", round(cv, 2), "%")
```

Q3. Build a Variance and Standard Deviation Calculator
Create a reusable Python program for the following datasets.

data_1 = [10, 20, 30, 40, 50]

data_2 = [20, 20, 20, 20, 20]

data_3 = [5, 10, 15, 20, 100]

data_4 = [12, 14, 16, 18, 20, 22]

(a) Implement separate functions to calculate the mean, population variance, sample variance, population standard deviation, and sample standard deviation without using NumPy's built-in statistical calculation functions. Correctly handle odd- and even-sized datasets.

(b) Test all four datasets and produce a formatted report. Identify the dataset with zero variability and the dataset most affected by an extreme observation. Justify your conclusions using the calculated variance and standard deviation.

```python
data_1 = [10, 20, 30, 40, 50]
data_2 = [20, 20, 20, 20, 20]
data_3 = [5, 10, 15, 20, 100]
data_4 = [12, 14, 16, 18, 20, 22]


def mean(data):
    return sum(data) / len(data)

def sample_variance(data):
    avg = mean(data)
    squared_sum = 0
    for x in data:
        squared_sum += (x - avg) ** 2
    return squared_sum / (len(data) - 1)

def sample_std(data):
    return sample_variance(data) ** 0.5
datasets = [data_1, data_2, data_3, data_4]
for i, data in enumerate(datasets, start=1):
    print("\nDataset", i)
    print("Mean:", mean(data))
    print("Sample Variance:", sample_variance(data))
    print("Sample Standard Deviation:", sample_std(data))
```

Q4. Investigating the Effect of Mathematical Transformations
A researcher records these measurements:

measurements = [12, 15, 18, 20, 25, 30, 40]
(a) Create two new datasets: one by adding 10 to every observation and another by multiplying every observation by 4. Calculate the mean, population variance, and population standard deviation for all three datasets. Use Python to verify the mathematical relationships between the original and transformed statistics.

(b) Create a fourth dataset by multiplying every observation by \(-2\). Calculate its mean, variance, and standard deviation. Explain why a negative multiplier changes the direction of the values but does not make variance or standard deviation negative.

```python
measurements = [12, 15, 18, 20, 25, 30, 40]

def calculate_mean(data):
    return sum(data) / len(data)

def calculate_variance(data):
    avg = calculate_mean(data)
    total = 0
    for x in data:
        total += (x - avg) ** 2
    return total / len(data)

def calculate_std(data):
    return calculate_variance(data) ** 0.5
original = measurements
added_10 = [x + 10 for x in measurements]
multiplied_4 = [x * 4 for x in measurements]
datasets = {"Original": original, "Add 10": added_10,"Multiply by 4": multiplied_4}

for name, data in datasets.items():
    print("\nDataset:", name)
    print("Values:", data)
    print("Mean:", calculate_mean(data))
    print("Population Variance:", calculate_variance(data))
    print("Population Standard Deviation:", calculate_std(data))
```

```python
multiplied_negative_2 = [x * -2 for x in measurements]
print("Original Data:", measurements)
print("New Data:", multiplied_negative_2)

print("Mean:", calculate_mean(multiplied_negative_2))
print("Population Variance:", calculate_variance(multiplied_negative_2))
print("Population Standard Deviation:", calculate_std(multiplied_negative_2))
```

Q5. Outlier Detection and Robust Spread Analysis
A data analyst records daily sales, in thousands of rupees.

sales = [    12, 14, 15, 15, 16, 17, 18,    19, 20, 21, 22, 24, 25, 120]
(a) Write Python code to calculate the mean, median, population variance, and population standard deviation. Then identify potential outliers using the IQR method and report the quartiles, IQR, lower bound, upper bound, and flagged observations.

(b) Temporarily exclude the flagged outlier and recalculate the mean, median, variance, and standard deviation. Compare the percentage changes in variance and standard deviation, and recommend which statistics best describe typical daily sales while explaining why an extreme observation should be investigated before removal.

```python
sales = [12, 14, 15, 15, 16, 17, 18,
         19, 20, 21, 22, 24, 25, 120]

def calculate_mean(data):
    return sum(data) / len(data)

def calculate_median(data):
    sorted_data = sorted(data)
    n = len(sorted_data)
    if n % 2 == 0:
        return (sorted_data[n // 2 - 1] +
                sorted_data[n // 2]) / 2
    else:
        return sorted_data[n // 2]

def calculate_variance(data):
    avg = calculate_mean(data)
    total = 0
    for x in data:
        total += (x - avg) ** 2
    return total / len(data)

def calculate_std(data):
    return calculate_variance(data) ** 0.5

def calculate_percentile(data, p):
    sorted_data = sorted(data)
    position = (len(sorted_data) - 1) * p

    lower_index = int(position)
    upper_index = lower_index + 1

    if upper_index >= len(sorted_data):
        return sorted_data[lower_index]

    fraction = position - lower_index

    return (sorted_data[lower_index] + fraction * (sorted_data[upper_index] - sorted_data[lower_index]))

mean = calculate_mean(sales)
median = calculate_median(sales)
variance = calculate_variance(sales)
std_dev = calculate_std(sales)

Q1 = calculate_percentile(sales, 0.25)
Q3 = calculate_percentile(sales, 0.75)

IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = []

for x in sales:
    if x < lower_bound or x > upper_bound:
        outliers.append(x)

print("Original Sales:", sales)
print("Mean:", mean)
print("Median:", median)
print("Population Variance:", variance)
print("Population Standard Deviation:", std_dev)

print("\nQ1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Flagged Outliers:", outliers)
```

```python
clean_sales = [x for x in sales if x not in outliers]
old_mean = calculate_mean(sales)
old_median = calculate_median(sales)
old_variance = calculate_variance(sales)
old_std = calculate_std(sales)
new_mean = calculate_mean(clean_sales)
new_median = calculate_median(clean_sales)
new_variance = calculate_variance(clean_sales)
new_std = calculate_std(clean_sales)
variance_change = (
    (new_variance - old_variance) / old_variance) * 100

std_change = ((new_std - old_std) / old_std) * 100

print("Before Removing Outlier")
print("Mean:", old_mean)
print("Median:", old_median)
print("Variance:", old_variance)
print("Standard Deviation:", old_std)

print("\nAfter Removing Outlier")
print("Mean:", new_mean)
print("Median:", new_median)
print("Variance:", new_variance)
print("Standard Deviation:", new_std)

print("\nPercentage Change")
print("Variance Change (%):", variance_change)
print("Standard Deviation Change (%):", std_change)
```
